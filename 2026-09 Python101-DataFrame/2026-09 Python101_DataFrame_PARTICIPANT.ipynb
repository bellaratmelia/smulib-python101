**FYI — `.query()` vs the `&` / `|` style:** both give the *same result*, so which you use is a trade-off, not a right/wrong choice.

- 👍 **Pros:** far more readable, especially with several conditions; you can write `and` / `or` in plain English, and there's no need to wrap each condition in `( )`.
- 👎 **Cons:** the condition is just a text string, so typos hide *inside* it instead of erroring clearly, and it's harder to build dynamically (e.g. when a column name is stored in a variable).

There's rarely one "correct" method in pandas, only trade-offs — knowing *why* you'd pick one is the part that's yours to bring (an AI will happily generate either without telling you which fits).

## Selecting by position: `.iloc`

So far we've selected by **label** — column names, and `.loc` with a True/False mask. Sometimes you instead want rows or columns by their **position** (the 1st, 2nd, 3rd...), regardless of their names. That's what plain number-slicing and `.iloc` are for.

Just like slicing a list, the end number is **exclusive**.

```python
# select rows by POSITION (like slicing a list). Rows 6,7,8,9,10 — the end (11) is excluded

```

```python
# .iloc[rows, columns] selects by POSITION for both.
# here: rows at positions 5–10, and the column at position 1 (the 2nd column)

```

**Rule of thumb:** `.loc` selects by **label** (column names / index values), while `.iloc` selects by **position** (0, 1, 2, ...). Mixing them up is a classic source of "why did I get the wrong column?" bugs — including in AI-generated code that guessed at your data's layout. We'll use `.iloc` again in the time-series section to drop an unnamed column.
